```python
import pandas as pd

# Define the file path
#file_path = r"C:\Users\bazrafka\Desktop\counting\DiscussionPaperData\Python\Ali_Data_kg_ha_updated.xlsx"
file_path = r"C:\Users\bazrafka\Desktop\counting\DiscussionPaperData\Python\Ali_Data_kg_ha_updated_with_rescaled_yields.xlsx"


# Load the excel file
df = pd.read_excel(file_path)

# Create the mapping dictionary for market classes
# Note: Numeric IDs are stored as strings to ensure matching regardless of data type
market_class_map = {
    "2115": "Kidney", "2116": "Yellow", "2201": "Black", "2202": "Navy",
    "2217": "Black", "2218": "Navy", "2219": "Kidney",
    "2402": "Black", "2403": "Navy", "2404": "Black",
    "2405": "GN & Pinto", "2406": "GN & Pinto",
    "2407": "Red & Pink", "2408": "Mixed",
    "2409": "Mixed", "2410": "Mixed", "2411": "Mixed", "2412": "Mixed",
    "2502": "Black", "2503": "Navy", "2504": "Black",
    "2505": "GN & Pinto", "2506": "GN & Pinto",
    "2507": "Red & Pink", "2508": "Red & Pink", "2512": "Mixed",
    "24116": "Black", "25104": "Black", "25105": "Kidney", 
    "25106": "Cran", "25111": "Black & Navy", "25112": "Pinto", 
    "25113": "Kidney", "25114": "Cran"
}

# Function to generate the new experiment string
def create_experiment_name(row):
    # Assuming 'Experiment Name' contains the numbers (e.g., 2115)
    # Adjust 'Experiment Name', 'Year', and 'Loc' if your column names differ
    exp_id = str(row['Experiment Name']).strip()
    class_name = market_class_map.get(exp_id, "Unknown")
    
    # Extract last two digits of year (e.g., 2022 -> 22)
    year_short = str(row['year'])[-2:]
    location = str(row['loc'])
    
    return f"{class_name}_{year_short}_{location}"

# Create the new column 'Experiment'
# We use axis=1 to process row by row
df['Experiment'] = df.apply(create_experiment_name, axis=1)

# Save the updated data back to an excel file
# This keeps all original columns and adds the new 'Experiment' column at the end
#output_path = r"C:\Users\bazrafka\Desktop\counting\DiscussionPaperData\Python\Ali_Data_kg_ha_updated_rename.xlsx"
output_path = r"C:\Users\bazrafka\Desktop\counting\DiscussionPaperData\Python\Ali_Data_kg_ha_updated_with_rescaled_yields_rename.xlsx"

df.to_excel(output_path, index=False)

print(f"Process complete. New file saved at: {output_path}")
```
