```python
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import scipy.stats as stats
import warnings

# Suppress warnings for a cleaner output
warnings.simplefilter(action='ignore', category=FutureWarning)

# Define the file path and load the dataset
file_path = r"C:\Users\bazrafka\Desktop\counting\DiscussionPaperData\Python\Ali_Data_kg_ha_updated.xlsx"
df = pd.read_excel(file_path)

# Standardize column names to ensure they match the expected strings
df.columns = df.columns.str.strip()
df = df.rename(columns={'Corrected_Yield': 'CorrectedYield', 'Uncorrected_Yield': 'UncorrectedYield'})

# Group the data by Experiment Name to process each trial individually
for exp_name, exp_data in df.groupby('Experiment Name'):
    
    # Prepare data by removing rows where either yield value is missing
    uncorr_data = exp_data['UncorrectedYield'].dropna()
    corr_data = exp_data['CorrectedYield'].dropna()
    
    # Only proceed if there is enough data to generate a meaningful plot
    if len(uncorr_data) > 3 and len(corr_data) > 3:
        # Create a figure with one row and two columns for the subplots
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
        
        # Generate the QQ plot for the Uncorrected Yield
        sm.qqplot(uncorr_data, line='45', fit=True, ax=ax1)
        ax1.set_title(f'Uncorrected Yield - {exp_name}')
        ax1.set_xlabel('Theoretical Quantiles')
        ax1.set_ylabel('Sample Quantiles')
        
        # Generate the QQ plot for the Corrected Yield
        sm.qqplot(corr_data, line='45', fit=True, ax=ax2)
        ax2.set_title(f'Corrected Yield - {exp_name}')
        ax2.set_xlabel('Theoretical Quantiles')
        ax2.set_ylabel('Sample Quantiles')
        
        # Adjust layout to prevent title and label overlapping
        plt.tight_layout()
        
        # Save each trial's plot as a high-resolution PNG file
        file_name = f"QQ_Plot_{exp_name}.png".replace("/", "_")
        plt.savefig(file_name, dpi=300)
        plt.close(fig)

print("QQ plot generation is complete. Individual images for each trial have been saved to your directory.")
```
